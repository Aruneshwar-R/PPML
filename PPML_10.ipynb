  To understand file handling operations in python by writing , reading and appending text data using different file modes, and to learn the basic usage of json loads() and dumps() functions for data serialization and deserialization


Description:


1.   open a text file in write mode
2.   write the text content and close the file
3.   open the same file in read mode
4.   read the file content using the readline function and close the file
5.   open the same file in append mode
6.   add few more text and close the file
7.   open the same file in read mode



```python
#Open a text file in 'Write' mode to write the given lines
f=open("/content/sample_data/textfile.txt", 'w')
f.write("first line\n")
f.write("Second line\n")
f.write("third line\n")
f.close()
#Open a text file in 'Read' mode to read and print the available lines readline function
k=open("/content/sample_data/textfile.txt", 'r')
print(k.readline())
print(k.readline())
print(k.readline())
k.close()
```

```python
#Open a text file in 'Append' mode to add/append new line inputs
m=open("/content/sample_data/textfile.txt", 'a')
m.write("fourth line\n")
m.write("fifth line\n")
m.close()
#Open a text file in 'Read' mode to read and print the available full text read function
p=open("/content/sample_data/textfile.txt", 'r')
print(p.read())
p.close()
```

```python
with open("/content/sample_data/textfile.txt",'r') as file:
 lines=file.readlines()
 print(lines)
```

```python
import json
# JSON string:
x = '{ "name":"John", "age":30, "city":"New York" }'
# Convert JSON string to Python dictionary
y = json.loads(x)
print(y["age"])
```

```python
#Python dictionary
data = {
"name": "John",
"age": 30,
"city": "New York"
}
#Convert Python dictionary to JSON string
json_data = json.dumps(data)
print(json_data)
```
