```python
import pandas as pd
df=pd.DataFrame()
print(df)
```

```python
emp=pd.Series(["parker","john","smith","williams"])
id=pd.Series([102,203,3098,114])
frame={'emp':emp,"id":id}
result=pd.DataFrame(frame)

print("series to Data frame")
print(result)
```

```python
print("\n Adding New Column")
result["Age"]=pd.Series([35,23,40,38])
print(result)
```

```python
print("nDeleting one column:\n")
del result['Age']
print(result)
```

```python
print("\nExtracting the third row:\n")
print(result.loc[2])
```

```python
print("\nSlice rows: \n", result [1:3])
```

```python
d2 = pd.DataFrame([['Dale', 123], ['Mark', 143]], columns=['Emp', 'ID'])
pd.concat([result, d2])
```
